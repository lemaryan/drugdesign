```python
from dockstring import load_target
from rdkit import Chem
```

```python
smiles = "C1CCC(C1)[C@@H](CC#N)N2C=C(C=N2)C3=C4C=NC4=NC=N3"
```

```python
mol = Chem.MolFromSmiles(smiles)
mol
```

```python
jak2 = load_target("JAK2")
score, aux = jak2.dock(smiles)

print(score)
```

```python
lck = load_target("LCK")
lck_score, lck_aux = lck.dock(smiles)

print(lck_score)
```

```python
# Dock against LCK
lck = load_target("LCK")
lck_score, lck_aux = lck.dock(smiles)

print("LCK:", lck_score)
```

```python
print("JAK2:", score)
print("LCK:", lck_score)
print("JAK2 - LCK:", score - lck_score)
```

## Docking Results

A test molecule was docked against both JAK2 and LCK using DOCKSTRING.

The docking scores are compared to examine whether the molecule shows stronger
predicted binding to JAK2 than LCK. Lower (more negative) DOCKSTRING docking
scores indicate more favorable predicted binding.

The difference `JAK2 - LCK` is used here as a simple exploratory measure of
selectivity. A negative value means the JAK2 docking score is more favorable
than the LCK docking score.
